## 7) Plotly (WebGL) — 빠른 GUI 같은 탐색 (zoom/pan/legend toggle)

matplotlib보다 훨씬 빠름. 마우스로 zoom/pan, legend 클릭으로 해당 데이터셋 toggle.
노트북 셀에서 inline 표시되며 streamlit 없이도 GUI 같은 인터랙션 가능.

```python
# 작은 grid (3 모델 × 2 데이터셋) — 가장 빠르고 가벼움
ex.view_plotly(
    models=['CPC','ECG-Founder','ST-MEM'],
    datasets=['ptbxl','zzu'],
    preset='orig',
)
```

```python
# 나이대별 plotly (legend 클릭으로 bin toggle)
ex.view_plotly(
    models=['CPC','ECG-Founder'],
    datasets=['ptbxl','zzu'],
    color_mode='age',
    age_bins='0,18,40,60,80,200',
)
```

```python
# 라벨 모드 plotly
ex.view_plotly(
    models=['CPC','ECG-Founder'],
    datasets=['chapman'],
    color_mode='label',
    labels=[('chapman','AFIB')],
)
```
